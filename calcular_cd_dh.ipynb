```python
from spatialmath.base import *
import sympy as sp
```

```python
q1,q2,L1,L2 = sp.symbols("q1 q2 L1 L2")
```

```python
def mdh(a,alpha,d,theta):
    return trotx(alpha) @ transl(a,0,0) @ trotz(theta) @ transl(0,0,d)

def simp(T):
    return sp.Matrix(T).applyfunc(sp.simplify)
    
```

```python
T01 = mdh(0,0,0,q1)
T12 = mdh(L1,0,0,q2)
T23 = mdh(L2,0,0,0)

CD = T01 @ T12 @ T23
```

```python
simp(CD)
```

```python
q1,q2,q3 = sp.symbols("q1 q2 q3")
d1,l1,l2 = sp.symbols("d1 l1 l2")  

T01 = mdh(0,0,d1,q1)
T12 = mdh(l1,0,0,q2)
T23 = mdh(l2,sp.pi,q3,0)
```

```python
simp(T01)
```

```python
simp(T12)
```

```python
simp(T23)
```

```python
simp(T01 @ T12 @ T23)
```

```python
#ejemplo RP

l1,q1,q2 = sp.symbols("l1 q1 q2")

T01 = mdh(0,0,l1,q1)
T12 = mdh(0,0,q2,0)
```

```python
simp(T01 @ T12)
```

```python
#ejemplo esferico

l1,l2,q1,q2,q3 = sp.symbols("l1 l2 q1 q2 q3")


# a,alpha,d,theta
T01 = mdh(0,0,l1,q1)
T12 = mdh(0,sp.pi/2,0,q2+sp.pi/2)
#sin considerar l2, da un resultado
T23 = mdh(l2,sp.pi/2,q3,0)
```

```python
simp(T01)
```

```python
simp(T12)
```

```python
simp(T23)
```

```python
T02 = simp(T01 @ T12)
T0E = simp(T01 @ T12 @ T23)
```

```python
sp.trigsimp(T0E)
```
